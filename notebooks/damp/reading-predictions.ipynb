# Reading a prediction table

This example shows how a notebook becomes part of the research record. All values below are **synthetic**. They demonstrate formatting and model comparison, not biological findings.

## 1. Start with aligned records

Compare the same candidates across models before calculating disagreement.

```python
from statistics import mean

# Synthetic values for a publishing demonstration, not measured MICs.
records = [
    {"id": "demo-01", "model_a": 0.81, "model_b": 0.78},
    {"id": "demo-02", "model_a": 0.74, "model_b": 0.32},
    {"id": "demo-03", "model_a": 0.29, "model_b": 0.41},
    {"id": "demo-04", "model_a": 0.92, "model_b": 0.87},
]
for row in records:
    row["gap"] = abs(row["model_a"] - row["model_b"])

print(f"Mean absolute gap: {mean(row['gap'] for row in records):.3f}")
```

## 2. Inspect individual disagreements

An average can conceal the cases worth inspecting. In this toy table, `demo-02` has the largest gap.

```python
from IPython.display import HTML, display

rows = "".join(
    f"<tr><td>{r['id']}</td><td>{r['model_a']:.2f}</td>"
    f"<td>{r['model_b']:.2f}</td><td>{r['gap']:.2f}</td></tr>"
    for r in records
)
display(HTML("<table><thead><tr><th>Candidate</th><th>Model A</th>"
             "<th>Model B</th><th>Absolute gap</th></tr></thead>"
             f"<tbody>{rows}</tbody></table>"))
```

## 3. What this comparison cannot tell us

Disagreement does not identify the more accurate model. A biological evaluation needs comparable endpoints, units, bacterial strains and independent measurements.

For MIC comparisons, specify the concentration units and assay context; for classifiers, check whether the scores are calibrated before interpreting their differences.
